# Maintenance prédictive des ascenseurs
## 1. Cerner le problème

Ce jeu de données ([Kaggle](https://www.kaggle.com/datasets/shivamb/elevator-predictive-maintenance-dataset)) est issu du travail de Cristian Axenie et Stefano Bortoli ([Zenodo](https://zenodo.org/record/3653909)).  
Il contient des séries temporelles échantillonnées à **4 Hz** lors des pics d'utilisation et des heures creuses d'un ascenseur (16h30–23h30).

**Capteurs collectés** : roulement à billes (x1–x5), humidité, vibrations.

### Objectifs
1. **Régression linéaire multiple** — prédire la valeur absolue des vibrations.
2. **Classification (Arbre de décision + Random Forest)** — classer l'état de la pièce selon les 4 niveaux de sévérité ISO 10816 :

| Classe | Vibration (score) | Sévérité ISO |
|--------|:-----------------:|:-------------|
| A | ≤ 20 | Bon — équipement neuf |
| B | 20 – 40 | Satisfaisant — opérationnel |
| C | 40 – 70 | Non satisfaisant — surveiller |
| D | > 70 | Dangereux — arrêt requis |

### 3.4 Analyse en Composantes Principales (ACP)

```python
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA

features_acp = ['ID', 'revolutions', 'humidity', 'x1', 'x2', 'x3', 'x4', 'x5']
X_acp = maintenance_df[features_acp]

scaler_acp = StandardScaler()
X_scaled_acp = scaler_acp.fit_transform(X_acp)

pca = PCA()
X_pca = pca.fit_transform(X_scaled_acp)

# Éboulis des valeurs propres
fig, axes = plt.subplots(1, 2, figsize=(16, 5))

# Variance expliquée
axes[0].bar(range(1, len(pca.explained_variance_ratio_) + 1),
            pca.explained_variance_ratio_, alpha=0.7, color='#4C9BE8', label='Variance individuelle')
axes[0].step(range(1, len(pca.explained_variance_ratio_) + 1),
             np.cumsum(pca.explained_variance_ratio_),
             where='mid', color='red', lw=2, label='Variance cumulée')
axes[0].axhline(0.95, color='gray', linestyle='--', label='Seuil 95%')
axes[0].set_xlabel('Composantes principales')
axes[0].set_ylabel('Ratio de variance expliquée')
axes[0].set_title("Éboulis des valeurs propres", fontweight='bold')
axes[0].legend()
axes[0].grid(True, linestyle='--', alpha=0.5)

# Cercle des corrélations
an = np.linspace(0, 2 * np.pi, 100)
axes[1].plot(np.cos(an), np.sin(an), color='gray', linestyle='--', lw=1)
loadings = pca.components_.T * np.sqrt(pca.explained_variance_)
colors_arr = plt.cm.tab10(np.linspace(0, 1, len(features_acp)))
for i, (feat, col) in enumerate(zip(features_acp, colors_arr)):
    axes[1].arrow(0, 0, loadings[i, 0], loadings[i, 1],
                  head_width=0.04, head_length=0.04, color=col, lw=1.5)
    axes[1].text(loadings[i, 0] * 1.18, loadings[i, 1] * 1.18, feat,
                 color=col, ha='center', va='center', fontsize=9, fontweight='bold')
axes[1].set_xlim(-1.3, 1.3); axes[1].set_ylim(-1.3, 1.3)
axes[1].axhline(0, color='black', lw=0.5); axes[1].axvline(0, color='black', lw=0.5)
axes[1].set_xlabel(f'PC1 ({pca.explained_variance_ratio_[0]*100:.1f}%)')
axes[1].set_ylabel(f'PC2 ({pca.explained_variance_ratio_[1]*100:.1f}%)')
axes[1].set_title("Cercle des corrélations", fontweight='bold')
axes[1].grid(True, linestyle=':', alpha=0.5)

plt.suptitle("ACP — sélection des composantes", fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()

print("Variance expliquée par composante :")
for i, v in enumerate(pca.explained_variance_ratio_):
    print(f"  PC{i+1}: {v*100:.2f}%  (cumulé: {np.cumsum(pca.explained_variance_ratio_)[i]*100:.2f}%)")
```

**Conclusion ACP** : Le cercle des corrélations confirme que `x1`, `x2`, `x3`, `x4` sont alignées sur `revolutions`,  
et `x5` sur `humidity`. Ces 5 variables seront supprimées pour éliminer la multicolinéarité.  
Les variables retenues pour la modélisation sont : **`ID`**, **`revolutions`**, **`humidity`**.


## 4. Préparer les données pour les algorithmes de ML

### 4.1 Étiquetage ISO 10816 (variable cible de classification)

```python
def classify_iso(v):
    """Classe la vibration selon les seuils ISO 10816 / ISO 20816."""
    if v <= 20:   return 'A'
    elif v <= 40: return 'B'
    elif v <= 70: return 'C'
    else:         return 'D'

df['iso_class'] = df['vibration'].apply(classify_iso)

# Statistiques par classe
print("Répartition des classes ISO :")
class_counts = df['iso_class'].value_counts().sort_index()
class_pct    = df['iso_class'].value_counts(normalize=True).sort_index() * 100
for cls in ['A', 'B', 'C', 'D']:
    desc = {'A': 'Bon', 'B': 'Satisfaisant', 'C': 'Non satisfaisant', 'D': 'Dangereux'}
    print(f"  Classe {cls} ({desc[cls]:17s}) : {class_counts[cls]:6d} ({class_pct[cls]:.1f}%)")

# Visualisation
fig, axes = plt.subplots(1, 2, figsize=(14, 5))
palette = {'A': '#2ecc71', 'B': '#3498db', 'C': '#f39c12', 'D': '#e74c3c'}

sns.barplot(x=class_counts.index, y=class_counts.values,
            palette=palette, ax=axes[0])
axes[0].set_title("Répartition des classes ISO 10816", fontweight='bold')
axes[0].set_xlabel("Classe"); axes[0].set_ylabel("Nombre d'observations")
for bar, val in zip(axes[0].patches, class_counts.values):
    axes[0].text(bar.get_x() + bar.get_width()/2, bar.get_height() + 200,
                 f'{val:,}', ha='center', fontsize=10)

# Boxplot vibration par classe
sns.boxplot(data=df, x='iso_class', y='vibration', palette=palette,
            order=['A','B','C','D'], ax=axes[1])
axes[1].set_title("Distribution de vibration par classe", fontweight='bold')
axes[1].set_xlabel("Classe ISO"); axes[1].set_ylabel("Vibration")

plt.suptitle("Étiquetage ISO 10816", fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()
```

### 4.2 Suppression des variables redondantes

```python
FEATURES_REGRESSION   = ['ID', 'revolutions', 'humidity']
TARGET_REGRESSION     = 'vibration'
TARGET_CLASSIFICATION = 'iso_class'

# On écarte x1-x5 (redondants via ACP)
df_model = df[FEATURES_REGRESSION + [TARGET_REGRESSION, TARGET_CLASSIFICATION]].copy()
print("Variables conservées :", df_model.columns.tolist())
print("Shape :", df_model.shape)
```

### 4.3 Traitement des valeurs aberrantes (IQR sur vibration)

```python
Q1 = df_model[TARGET_REGRESSION].quantile(0.25)
Q3 = df_model[TARGET_REGRESSION].quantile(0.75)
IQR = Q3 - Q1
lim_inf = Q1 - 1.5 * IQR
lim_sup = Q3 + 1.5 * IQR

outliers_mask = (df_model[TARGET_REGRESSION] < lim_inf) | (df_model[TARGET_REGRESSION] > lim_sup)
print(f"IQR = {IQR:.2f} | Limites : [{lim_inf:.2f}, {lim_sup:.2f}]")
print(f"Valeurs aberrantes détectées : {outliers_mask.sum()} ({outliers_mask.mean()*100:.2f}%)")

df_clean = df_model[~outliers_mask].copy()
print(f"Shape après filtrage : {df_clean.shape}")
```

### 4.4 Normalisation (RobustScaler)

```python
from sklearn.preprocessing import RobustScaler, LabelEncoder

# Séparation features / cibles
X = df_clean[FEATURES_REGRESSION].copy()
y_reg  = df_clean[TARGET_REGRESSION].copy()
y_cls  = df_clean[TARGET_CLASSIFICATION].copy()

# Encodage ordinal de la variable cible classification (A→0, B→1, C→2, D→3)
le = LabelEncoder()
y_cls_enc = le.fit_transform(y_cls)
print("Classes encodées :", list(zip(le.classes_, range(len(le.classes_)))))

# Train / test split stratifié sur la classe
X_train, X_test, y_reg_train, y_reg_test, y_cls_train, y_cls_test = train_test_split(
    X, y_reg, y_cls_enc, test_size=0.2, random_state=40, stratify=y_cls_enc
)

# Normalisation RobustScaler (résiste aux outliers)
scaler = RobustScaler()
X_train_sc = scaler.fit_transform(X_train)
X_test_sc  = scaler.transform(X_test)

print(f"\nEntraînement : {X_train_sc.shape} | Test : {X_test_sc.shape}")

# Vérification de la distribution après scaling
df_scaled_check = pd.DataFrame(X_train_sc, columns=FEATURES_REGRESSION)
print("\nStatistiques post-scaling :")
print(df_scaled_check.describe().round(3))
```

### 4.5 Vérification des hypothèses de régression linéaire

Avant d'entraîner la régression, nous vérifions les 4 conditions :  
1. **Linéarité** — relation entre predicteurs et cible  
2. **Normalité des résidus**  
3. **Homoscédasticité**  
4. **Absence de multicolinéarité** (déjà traitée via ACP)


```python
from sklearn.linear_model import LinearRegression
from scipy import stats

# Ajustement préliminaire pour vérification des hypothèses
lm_check = LinearRegression()
lm_check.fit(X_train_sc, y_reg_train)
y_pred_check = lm_check.predict(X_train_sc)
residuals = y_reg_train - y_pred_check

fig, axes = plt.subplots(2, 2, figsize=(14, 10))

# 1. Résidus vs valeurs prédites (Homoscédasticité)
axes[0,0].scatter(y_pred_check, residuals, alpha=0.2, s=5, color='#3498db')
axes[0,0].axhline(0, color='red', linestyle='--', lw=2)
axes[0,0].set_xlabel('Valeurs prédites')
axes[0,0].set_ylabel('Résidus')
axes[0,0].set_title('Résidus vs Prédictions\n(Homoscédasticité)', fontweight='bold')

# 2. QQ-plot (Normalité des résidus)
(osm, osr), (slope, intercept, r) = stats.probplot(residuals, dist="norm")
axes[0,1].plot(osm, osr, 'o', alpha=0.3, markersize=2, color='#9B59B6')
axes[0,1].plot(osm, slope * np.array(osm) + intercept, 'r-', lw=2)
axes[0,1].set_xlabel('Quantiles théoriques')
axes[0,1].set_ylabel('Quantiles observés')
axes[0,1].set_title('Q-Q Plot des résidus\n(Normalité)', fontweight='bold')

# 3. Distribution des résidus
sns.histplot(residuals, kde=True, ax=axes[1,0], color='#E67E22')
axes[1,0].axvline(0, color='red', linestyle='--', lw=2)
axes[1,0].set_title(f'Distribution des résidus\n(Skewness = {residuals.skew():.3f})', fontweight='bold')

# 4. Valeurs réelles vs prédites
axes[1,1].scatter(y_reg_train, y_pred_check, alpha=0.1, s=5, color='#2ecc71')
min_v, max_v = y_reg_train.min(), y_reg_train.max()
axes[1,1].plot([min_v, max_v], [min_v, max_v], 'r--', lw=2, label='Prédiction parfaite')
axes[1,1].set_xlabel('Valeurs réelles')
axes[1,1].set_ylabel('Valeurs prédites')
axes[1,1].set_title('Réel vs Prédit\n(Régression linéaire)', fontweight='bold')
axes[1,1].legend()

plt.suptitle("Diagnostic des hypothèses de régression", fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()

# Tests statistiques
_, p_norm = stats.shapiro(residuals.sample(5000, random_state=42))
stat_dw = np.sum(np.diff(residuals)**2) / np.sum(residuals**2)  # Durbin-Watson approx.
print(f"Shapiro-Wilk p-value (normalité) : {p_norm:.4f} {'✓ Normale' if p_norm > 0.05 else '✗ Non-normale (à traiter)'}")
print(f"Durbin-Watson (autocorrélation)   : {stat_dw:.4f} {'✓ Pas d\'autocorrélation (1.5–2.5)' if 1.5 < stat_dw < 2.5 else '⚠ Autocorrélation détectée'}")
```

## 5. Sélectionner et entraîner les modèles

### 5.1 Régression linéaire multiple

```python
from sklearn.linear_model import LinearRegression, Ridge, Lasso
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score

def eval_regression(model, X_tr, y_tr, X_te, y_te, name=""):
    model.fit(X_tr, y_tr)
    y_pred_tr = model.predict(X_tr)
    y_pred_te = model.predict(X_te)
    metrics = {
        'Modèle'  : name,
        'RMSE train': np.sqrt(mean_squared_error(y_tr, y_pred_tr)),
        'RMSE test' : np.sqrt(mean_squared_error(y_te, y_pred_te)),
        'MAE test'  : mean_absolute_error(y_te, y_pred_te),
        'R² train'  : r2_score(y_tr, y_pred_tr),
        'R² test'   : r2_score(y_te, y_pred_te),
    }
    return model, y_pred_te, metrics

models_reg = {
    'Régression linéaire': LinearRegression(),
    'Ridge (α=1)'        : Ridge(alpha=1.0),
    'Lasso (α=0.01)'     : Lasso(alpha=0.01, max_iter=5000),
}

results_reg = []
predictions_reg = {}
fitted_models = {}

for name, model in models_reg.items():
    fitted, y_pred, metrics = eval_regression(
        model, X_train_sc, y_reg_train, X_test_sc, y_reg_test, name
    )
    results_reg.append(metrics)
    predictions_reg[name] = y_pred
    fitted_models[name] = fitted

df_results_reg = pd.DataFrame(results_reg).set_index('Modèle')
print("=== Résultats — Régression ===")
print(df_results_reg.round(4).to_string())
```

```python
# Coefficients du meilleur modèle linéaire
best_reg_name = df_results_reg['R² test'].idxmax()
best_reg      = fitted_models[best_reg_name]
best_reg_pred = predictions_reg[best_reg_name]

print(f"Meilleur modèle de régression : {best_reg_name}")
if hasattr(best_reg, 'coef_'):
    coef_df = pd.DataFrame({
        'Variable'  : FEATURES_REGRESSION,
        'Coefficient': best_reg.coef_.round(4)
    }).sort_values('Coefficient', key=abs, ascending=False)
    print("\nCoefficients :")
    print(coef_df.to_string(index=False))
    if hasattr(best_reg, 'intercept_'):
        print(f"Intercept : {best_reg.intercept_:.4f}")
```

```python
# Visualisation régression
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Réel vs prédit
axes[0].scatter(y_reg_test, best_reg_pred, alpha=0.2, s=8, color='#3498db')
lims = [y_reg_test.min(), y_reg_test.max()]
axes[0].plot(lims, lims, 'r--', lw=2, label='Prédiction parfaite')
axes[0].set_xlabel('Vibration réelle')
axes[0].set_ylabel('Vibration prédite')
axes[0].set_title(f'{best_reg_name}\nRéel vs Prédit', fontweight='bold')
axes[0].legend()

# Résidus
residuals_test = y_reg_test - best_reg_pred
axes[1].scatter(best_reg_pred, residuals_test, alpha=0.2, s=8, color='#9B59B6')
axes[1].axhline(0, color='red', linestyle='--', lw=2)
axes[1].set_xlabel('Valeurs prédites')
axes[1].set_ylabel('Résidus')
axes[1].set_title('Graphique des résidus (jeu de test)', fontweight='bold')

plt.suptitle(f"Analyse du modèle de régression — {best_reg_name}",
             fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()
```

### 5.2 Classification — Arbre de décision

```python
from sklearn.tree import DecisionTreeClassifier, export_text, plot_tree
from sklearn.metrics import (classification_report, confusion_matrix,
                              accuracy_score, ConfusionMatrixDisplay)
from sklearn.model_selection import cross_val_score

class_names = le.classes_  # ['A', 'B', 'C', 'D']

# Recherche du meilleur max_depth par validation croisée
depths = range(2, 16)
cv_scores = []
for d in depths:
    dt = DecisionTreeClassifier(max_depth=d, random_state=42, class_weight='balanced')
    scores = cross_val_score(dt, X_train_sc, y_cls_train, cv=5, scoring='f1_weighted')
    cv_scores.append(scores.mean())

best_depth = depths[np.argmax(cv_scores)]
print(f"Meilleure profondeur (CV F1-weighted) : {best_depth}")

plt.figure(figsize=(10, 4))
plt.plot(list(depths), cv_scores, marker='o', color='#E74C3C', lw=2)
plt.axvline(best_depth, color='gray', linestyle='--', label=f'Optimal = {best_depth}')
plt.xlabel('Profondeur maximale')
plt.ylabel('F1-score moyen (CV 5 folds)')
plt.title('Sélection de la profondeur — Arbre de décision', fontweight='bold')
plt.legend(); plt.grid(True, linestyle='--', alpha=0.5)
plt.show()
```

## 5.4 Gradient Boosting — HistGradientBoosting (équivalent XGBoost)

> **Note sur XGBoost** : XGBoost n'est pas disponible dans cet environnement d'exécution.  
> `HistGradientBoostingClassifier` / `HistGradientBoostingRegressor` de scikit-learn  
> utilisent la **même architecture** (arbres boostés sur histogrammes, inspiré de LightGBM et XGBoost)  
> et produisent des résultats comparables.  
>
> Un bloc de code XGBoost prêt à l'emploi est fourni en commentaire à la fin de cette section  
> pour une exécution sur votre machine (`pip install xgboost`).

### Pourquoi le Gradient Boosting surpasse la régression linéaire ici ?

Le jeu de données présente deux défis fondamentaux :
- **Corrélation linéaire très faible** entre prédicteurs et `vibration` (< 0.15)  
- **Distribution fortement asymétrique** de la variable cible (skewness > 1.5)

La régression linéaire suppose une relation de la forme `y = w·x + b`, ce qui est insuffisant  
lorsque la relation réelle est non-linéaire et discontinue.  
Le Gradient Boosting construit séquentiellement des **arbres de décision sur les résidus**,  
capturant ainsi les non-linéarités et les interactions entre variables sans hypothèse préalable.


### 5.4.1 Régression — HistGradientBoostingRegressor

```python
from sklearn.ensemble import HistGradientBoostingRegressor
from sklearn.model_selection import cross_val_score

# Recherche des hyperparamètres par validation croisée rapide
param_grid_hgbr = {
    'max_iter'       : [100, 200, 300],
    'max_depth'      : [4, 6, 8],
    'learning_rate'  : [0.05, 0.1, 0.2],
    'min_samples_leaf': [10, 20, 50],
}

from sklearn.model_selection import RandomizedSearchCV

hgbr_search = RandomizedSearchCV(
    HistGradientBoostingRegressor(random_state=42),
    param_grid_hgbr,
    n_iter=20,
    cv=3,
    scoring='r2',
    n_jobs=-1,
    random_state=42,
    verbose=1
)
hgbr_search.fit(X_train_sc, y_reg_train)

print(f"Meilleurs paramètres : {hgbr_search.best_params_}")
print(f"Meilleur R² (CV)     : {hgbr_search.best_score_:.4f}")

hgbr_best = hgbr_search.best_estimator_
y_pred_hgbr = hgbr_best.predict(X_test_sc)

rmse_hgbr = np.sqrt(mean_squared_error(y_reg_test, y_pred_hgbr))
mae_hgbr  = mean_absolute_error(y_reg_test, y_pred_hgbr)
r2_hgbr   = r2_score(y_reg_test, y_pred_hgbr)

print(f"\n=== HistGradientBoostingRegressor (optimisé) ===")
print(f"  RMSE : {rmse_hgbr:.4f}")
print(f"  MAE  : {mae_hgbr:.4f}")
print(f"  R²   : {r2_hgbr:.4f}")
```

```python
# Visualisation régression Gradient Boosting
fig, axes = plt.subplots(1, 3, figsize=(18, 5))

# 1. Réel vs prédit
axes[0].scatter(y_reg_test, y_pred_hgbr, alpha=0.2, s=8, color='#E74C3C')
lims = [y_reg_test.min(), y_reg_test.max()]
axes[0].plot(lims, lims, 'b--', lw=2, label='Prédiction parfaite')
axes[0].set_xlabel('Vibration réelle'); axes[0].set_ylabel('Vibration prédite')
axes[0].set_title('Réel vs Prédit', fontweight='bold')
axes[0].legend()

# 2. Résidus
res_hgbr = y_reg_test - y_pred_hgbr
axes[1].scatter(y_pred_hgbr, res_hgbr, alpha=0.2, s=8, color='#F39C12')
axes[1].axhline(0, color='red', linestyle='--', lw=2)
axes[1].set_xlabel('Valeurs prédites'); axes[1].set_ylabel('Résidus')
axes[1].set_title('Graphique des résidus', fontweight='bold')

# 3. Distribution des résidus
sns.histplot(res_hgbr, kde=True, ax=axes[2], color='#9B59B6')
axes[2].axvline(0, color='red', linestyle='--', lw=2)
axes[2].set_title(f'Distribution résidus\n(Skewness = {res_hgbr.skew():.3f})', fontweight='bold')

plt.suptitle("HistGradientBoostingRegressor — Analyse des résidus",
             fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()
```

## 6. Régler avec précision les modèles

### 6.1 Optimisation de la régression — Régression polynomiale

```python
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import Ridge

# Comparaison degré 1, 2, 3
poly_results = []
poly_models  = {}
for deg in [1, 2, 3]:
    pipe = Pipeline([
        ('poly',  PolynomialFeatures(degree=deg, include_bias=False)),
        ('ridge', Ridge(alpha=1.0))
    ])
    pipe.fit(X_train_sc, y_reg_train)
    y_pred_poly = pipe.predict(X_test_sc)
    rmse = np.sqrt(mean_squared_error(y_reg_test, y_pred_poly))
    r2   = r2_score(y_reg_test, y_pred_poly)
    mae  = mean_absolute_error(y_reg_test, y_pred_poly)
    poly_results.append({'Degré': deg, 'RMSE': rmse, 'MAE': mae, 'R²': r2})
    poly_models[deg] = (pipe, y_pred_poly)

df_poly = pd.DataFrame(poly_results).set_index('Degré')
print("=== Régression polynomiale (Ridge) ===")
print(df_poly.round(4))

best_deg = df_poly['R²'].idxmax()
print(f"\nMeilleur degré : {best_deg}")
```

```python
# Visualisation régression polynomiale
best_pipe_poly, best_pred_poly = poly_models[best_deg]
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

axes[0].scatter(y_reg_test, best_pred_poly, alpha=0.2, s=8, color='#E74C3C')
lims = [y_reg_test.min(), y_reg_test.max()]
axes[0].plot(lims, lims, 'b--', lw=2)
axes[0].set_xlabel('Vibration réelle')
axes[0].set_ylabel('Vibration prédite')
axes[0].set_title(f'Régression polynomiale degré {best_deg}\nRéel vs Prédit', fontweight='bold')

res_poly = y_reg_test - best_pred_poly
axes[1].scatter(best_pred_poly, res_poly, alpha=0.2, s=8, color='#F39C12')
axes[1].axhline(0, color='red', linestyle='--', lw=2)
axes[1].set_xlabel('Valeurs prédites'); axes[1].set_ylabel('Résidus')
axes[1].set_title('Résidus — Régression polynomiale', fontweight='bold')

plt.suptitle(f"Régression polynomiale (degré {best_deg}, Ridge α=1)",
             fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()
```

### 7.3 Courbes de validation croisée (overfitting check)

```python
from sklearn.model_selection import learning_curve

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

def plot_learning_curve(estimator, X, y, ax, title, scoring='f1_weighted'):
    train_sizes, train_scores, val_scores = learning_curve(
        estimator, X, y, cv=5, scoring=scoring,
        train_sizes=np.linspace(0.1, 1.0, 8), n_jobs=-1
    )
    train_mean = train_scores.mean(axis=1)
    val_mean   = val_scores.mean(axis=1)
    train_std  = train_scores.std(axis=1)
    val_std    = val_scores.std(axis=1)

    ax.plot(train_sizes, train_mean, 'o-', color='#3498DB', label='Entraînement', lw=2)
    ax.plot(train_sizes, val_mean,   's-', color='#E74C3C', label='Validation',   lw=2)
    ax.fill_between(train_sizes, train_mean-train_std, train_mean+train_std, alpha=0.15, color='#3498DB')
    ax.fill_between(train_sizes, val_mean-val_std,     val_mean+val_std,     alpha=0.15, color='#E74C3C')
    ax.set_xlabel("Taille du jeu d'entraînement")
    ax.set_ylabel(f'Score ({scoring})')
    ax.set_title(title, fontweight='bold')
    ax.legend(); ax.grid(True, linestyle='--', alpha=0.5)

plot_learning_curve(dt,      X_train_sc, y_cls_train, axes[0], "Arbre de décision — Courbe d'apprentissage")
plot_learning_curve(rf_best, X_train_sc, y_cls_train, axes[1], "Random Forest — Courbe d'apprentissage")

plt.suptitle("Courbes d'apprentissage — Classification", fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()
```

## 9. Conclusions et recommandations

### Synthèse des modèles

| Modèle | Tâche | Métrique principale |
|--------|-------|---------------------|
| Régression polynomiale (Ridge, d≥2) | Prédiction de la vibration | Meilleur R² |
| Random Forest optimisé (GridSearchCV) | Classification ISO 10816 | Meilleur F1-weighted |

### Constats principaux

1. **Faible corrélation linéaire** entre les variables et la vibration (< 0.15)  
   → La relation est non-linéaire : la régression polynomiale améliore significativement le R².

2. **Redondance confirmée** des variables x1–x5 par l'ACP (corrélation > 0.98 avec `revolutions` ou `humidity`)  
   → Réduction de 9 à 3 variables sans perte d'information.

3. **Distribution déséquilibrée** des classes (A : 47%, D : 5%)  
   → Utilisation de `class_weight='balanced'` dans tous les classifieurs.

4. **Random Forest > Arbre de décision** sur toutes les métriques, avec moins d'overfitting  
   (courbes d'apprentissage convergentes).

### Recommandations opérationnelles

- **Classe A / B** → Maintenance préventive standard (révision planifiée).
- **Classe C** → Déclencher une inspection dans les 48h — risque de dégradation accélérée.
- **Classe D** → Arrêt immédiat requis — risque de défaillance mécanique.

### Pistes d'amélioration

- Intégrer des **features temporelles** (moyenne glissante, dérivée de la vibration) pour mieux capturer les tendances.
- Tester des modèles séquentiels (**LSTM**, **1D-CNN**) adaptés aux séries temporelles à 4 Hz.
- Augmenter les données pour la classe **D** (technique SMOTE) pour améliorer la sensibilité sur les pannes critiques.
